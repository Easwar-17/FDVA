```python
import pandas as pd

# Load the dataset
df = pd.read_csv('file:///C:/Users/HDC0422222/Documents/New%20folder%20(2)/sample_data_set.csv.txt')

# Clean column names by stripping trailing/leading whitespaces
df.columns = df.columns.str.strip()

# 1. Viewing and Inspecting Data
print("--- FIRST 5 ROWS ---")
print(df.head())

print("\n--- LAST 5 ROWS ---")
print(df.tail())

print("\n--- DATASET INFO ---")
print(df.info())

print("\n--- SUMMARY STATISTICS ---")
print(df.describe())

print("\n--- COLUMN NAMES ---")
print(df.columns)

# 2. Inspecting Categorical Columns
categorical_cols = [
    'Gender',
    'Married',
    'Dependents',
    'Education',
    'Self_Employed',
    'Property_Area',
]
for col in categorical_cols:
  if col in df.columns:
    unique_vals = df[col].unique()
    print(f"\nUnique values in '{col}': {unique_vals}")

# 3. Data Cleaning / Type Conversion for Dependents
if 'Dependents' in df.columns:
  df['Dependents'] = df['Dependents'].replace('3+', 3)
  df['Dependents'] = pd.to_numeric(df['Dependents'], errors='coerce').astype(
      'Int64'
  )

  print("\nValue counts in 'Dependents' after conversion:")
  print(df['Dependents'].value_counts(dropna=False))

# 4. Measures of Central Tendency & Dispersion
cols = ['ApplicantIncome', 'CoapplicantIncome', 'LoanAmount']
for col in cols:
  if col in df.columns:
    print(f"\nStatistics for '{col}':")
    print(f"Mean: {df[col].mean():.2f}")
    print(f"Median: {df[col].median():.2f}")
    print(f"Mode: {df[col].mode().values}")
    print(f"Range: {df[col].max() - df[col].min():.2f}")
    print(f"Variance: {df[col].var():.2f}")
    print(f"Standard Deviation: {df[col].std():.2f}")
```
